# First , Let´s find out the brands that are missing out (only one brand manufacter that car model)

```python
df[df['brand'].isnull()]
```

```python
# set to nan various possible value in dataset
df['brand'] = df['brand'].replace(['null', 'NA', 'N/A', 'NaN', ' ', None], np.nan)
df['model'] = df['model'].replace(['null', 'NA', 'N/A', 'NaN', ' ', None], np.nan)

df[df['brand'].isnull()]
```

```python
# Filter lines where brand and models are not NaN
df_valid = df.dropna(subset=['brand', 'model'])

#  Remove duplicates , keeping a single one
df_unique = df_valid.drop_duplicates(subset=['model'])

# Dict: model -> brand
map_brands = df_unique.set_index('model')['brand'].to_dict()

# Use model as dict key and fill brand
df['brand'] = df['brand'].fillna(df['model'].map(map_brands))

print(df.isnull().sum())
```

# Cars with no Models or Color are going to be dropped #
- less than 1% of the dataset

```python
df = df.dropna(subset=['color'])
df = df.dropna(subset=['model'])
print(df.isnull().sum())
```

# Mileage and year --> statistic aproach

- Mileage first

```python
# 1. Ensure zeros are treated as nulls
df['mileage'] = df['mileage'].replace([0, '0', ''], np.nan)

# 2. Create Price Bins
df['price_bracket'] = pd.qcut(df['price'], q=10, labels=False, duplicates='drop')

# 3. Grouping by Brand, Model, and Price Bracket. bmp_group
bmp_group = ['brand', 'model', 'price_bracket']
median_specific = df.groupby(bmp_group)['mileage'].transform('median')

# Fill the missing values with our highly specific median
df['mileage'] = df['mileage'].fillna(median_specific)
```

```python
print(df.isnull().sum())
```

```python
#Droping 3 left (less than 1%)
df = df.dropna(subset=['mileage'])
print(df.isnull().sum())
```

- Year inference

```python
# 1. Ensure zeros are treated as nulls
df['year'] = df['year'].replace([0, '0', ''], np.nan)

# 2. Grouping by Brand, Model, and Price Bracket. bmp_group
bmp_group = ['brand', 'model', 'price_bracket']
median_specific = df.groupby(bmp_group)['year'].transform('median')

# 3. Fill the missing values with our highly specific median
df['year'] = df['year'].fillna(median_specific)
```

```python
print(df.isnull().sum())
```

```python
# 2 cars left, we are going to drop
df = df.dropna(subset=['year'])
print(df.isnull().sum())
```

# 79 prices unknow, 3 methods (XGBoost, Randon Forest and Linear Regression )

## ML Setup — Feature Engineering & Train/Predict Split

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score
from xgboost import XGBRegressor

FEATURES = ["brand", "model", "year", "mileage", "color", "title_status", "state", "condition"]

df_ml = df[FEATURES + ["price"]].copy()

# Encode all categorical columns as integers
encoders = {}
for col in ["brand", "model", "color", "title_status", "state", "condition"]:
    le = LabelEncoder()
    df_ml[col] = le.fit_transform(df_ml[col].astype(str))
    encoders[col] = le

# Split: rows with price = train | rows without price = predict
train   = df_ml[df_ml["price"].notna()]
predict = df_ml[df_ml["price"].isna()]

X_train   = train[FEATURES]
y_train   = train["price"]
X_predict = predict[FEATURES]

print(f"Train: {len(X_train)} rows | To predict: {len(X_predict)} rows")
```

## Model 1 — XGBoost

```python
xgb_model = XGBRegressor(n_estimators=300, learning_rate=0.05, max_depth=6, random_state=42, verbosity=0)
xgb_model.fit(X_train, y_train)

y_pred_xgb_train = xgb_model.predict(X_train)
prices_xgb = xgb_model.predict(X_predict)

mae_xgb = mean_absolute_error(y_train, y_pred_xgb_train)
r2_xgb  = r2_score(y_train, y_pred_xgb_train)
print(f"XGBoost  | MAE: ${mae_xgb:,.0f}  |  R2: {r2_xgb:.4f}")
```

## Model 2 — Random Forest

```python
rf_model = RandomForestRegressor(n_estimators=300, random_state=42, n_jobs=-1)
rf_model.fit(X_train, y_train)

y_pred_rf_train = rf_model.predict(X_train)
prices_rf = rf_model.predict(X_predict)

mae_rf = mean_absolute_error(y_train, y_pred_rf_train)
r2_rf  = r2_score(y_train, y_pred_rf_train)
print(f"Random Forest  | MAE: ${mae_rf:,.0f}  |  R2: {r2_rf:.4f}")
```

## Model 3 — Linear Regression

```python
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)

y_pred_lr_train = lr_model.predict(X_train)
prices_lr = lr_model.predict(X_predict)

mae_lr = mean_absolute_error(y_train, y_pred_lr_train)
r2_lr  = r2_score(y_train, y_pred_lr_train)
print(f"Linear Regression  | MAE: ${mae_lr:,.0f}  |  R2: {r2_lr:.4f}")
```

## Comparison — XGBoost vs Random Forest vs Linear Regression

```python
# Summary table
summary = pd.DataFrame({
    "Model":      ["XGBoost", "Random Forest", "Linear Regression"],
    "MAE (train)": [mae_xgb, mae_rf, mae_lr],
    "R2 (train)":  [r2_xgb,  r2_rf,  r2_lr]
})
print(summary.to_string(index=False))

# Side-by-side predictions for the missing prices
result = df.loc[X_predict.index, ["brand", "model", "year", "mileage", "color", "state"]].copy()
result["XGBoost"]           = prices_xgb.round(0)
result["Random Forest"]     = prices_rf.round(0)
result["Linear Regression"] = prices_lr.round(0)
result
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

models = ["XGBoost", "Random Forest", "Linear Regression"]
maes   = [mae_xgb, mae_rf, mae_lr]
r2s    = [r2_xgb,  r2_rf,  r2_lr]

axes[0].bar(models, maes, color=["steelblue", "seagreen", "tomato"])
axes[0].set_title("MAE  — lower is better")
axes[0].set_ylabel("Mean Absolute Error ($)")
for i, v in enumerate(maes):
    axes[0].text(i, v + 50, f"${v:,.0f}", ha="center", fontsize=10)

axes[1].bar(models, r2s, color=["steelblue", "seagreen", "tomato"])
axes[1].set_title("R2 Score — higher is better")
axes[1].set_ylabel("R2")
axes[1].set_ylim(0, 1.05)
for i, v in enumerate(r2s):
    axes[1].text(i, v + 0.01, f"{v:.4f}", ha="center", fontsize=10)

plt.tight_layout()
plt.show()
```
